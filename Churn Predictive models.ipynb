```python
import pandas as pd
df = pd.read_excel(r"C:\Users\Sopuluchi\Documents\Tech\Power BI\Churn Analysis\New folder\Churn Analysis using ML Predictive Models.xlsx")

from catboost import CatBoostClassifier
```

```python
x = df.drop(columns = "Is_churned")                                            # input feature used for training
```

```python
y = df["Is_churned"]                                                           # target column the model will predictions
```

```python
cat_feat = x.select_dtypes(include = ["object", "string"]).columns.tolist()    # identify categorical columns & convert them to list
```

```python
model = CatBoostClassifier(iterations = 500, learning_rate = 0.05, depth = 6, verbose = 0)
                                                                               # create your model, that is a new instance of CatBoostClassifier
```

```python
model.fit(x, y, cat_features= cat_feat)                                        # training the model
```

```python
New_subscriber = pd.DataFrame({"Gender": ["Male", "Female"], "Age": [ 20, 53], "Genre": ["Classical", "HipHop"]})
                                                                               # New features our model will predict
```

```python
predictions = model.predict(New_subscriber)                                    # making prediction
```

```python
print(predictions)                                                       # printing the result of our prediction
```
