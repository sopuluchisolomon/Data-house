```python
df = df.drop_duplicates()                                                              # removing duplicate row
```

```python
df = df.drop(columns = "Not Useful Column")                                            # dropping Not Useful Column column
```

```python
df["Name"] = df["Name"].str.replace(r"[/.?,\\]", " ", regex = True)                    # removing unwanted characters everywhere
```

```python
df[["First Name", "Last Name"]] = df["Name"].str.split(expand = True, n = 1)           # spliting "Name" column to 
```

```python
cols = df.columns.tolist()                                                             # re-arrangeing the columns position
cols.remove("First Name")
cols.remove("Last Name")

cols = [cols[0], "First Name", "Last Name"] + cols[1:]
df = df[cols]
```

```python
df = df.drop(columns ="Name")                                                          # dropping "Name" column
```

```python
df["Gender"] = df["Gender"].replace({"Mal": "Male", "M": "Male", "ma": "Male", "mal": "Male", "F": "Female", "fe": "Female"})
                                                                                       # replace "Male" & "Female" Gender with corresponding string
```

```python
df["Phone Number"] = df["Phone Number"].str.replace("[^a-zA-Z0-9]", "", regex = True)  # dropping characters in the "Phone Number" column
```

```python
df["Phone Number"] = df["Phone Number"].apply(lambda x: str(x))                        # Converting "Phone Number"column to string
```

```python
df["Phone Number"] = df["Phone Number"].apply(lambda x: x[0:3]+ "-"+ x[3:6]+ "-"+ x[6:10])
                                                                                       # slicing through "phone Number" elements
```

```python
df["Phone Number"] = df["Phone Number"].str.replace({"nan--": " ", "Na--": " "})       # Removing letters from "phone Number" column
```
